Notebook auxiliar para descargar la versión más reciente de **Predict Future Sales** desde Kaggle. La cuenta debe tener acceso a la competencia y estar autenticada localmente. No incluyas tokens ni credenciales en este notebook.

## 1. Instalar e importar KaggleHub

La dependencia se instala en el entorno activo si aún no está disponible; luego se importa para descargar la competencia.

```python
%pip install kagglehub

import kagglehub
```

## 2. Descargar la versión más reciente

KaggleHub resuelve y descarga la versión más reciente disponible para la competencia indicada.

```python
from pathlib import Path

project_root = Path.cwd()
if not (project_root / "data" / "raw").exists() and (project_root.parent / "data" / "raw").exists():
    project_root = project_root.parent

raw_dir = project_root / "data" / "raw"
raw_dir.mkdir(parents=True, exist_ok=True)
path = kagglehub.competition_download(
    "competitive-data-science-predict-future-sales",
    output_dir=str(raw_dir),
)
print("Path to competition files:", path)
```

## 3. Mostrar la ruta de descarga

La variable `path` contiene la ruta que KaggleHub devolvió. El bloque de descarga ya la imprime; esta celda permite consultarla de nuevo sin repetir la descarga.

```python
print("Path to competition files:", path)
```
